## Train OntoVAE model

```python
# import packages
import sys
import scanpy as sc
sys.path.append('/workspace')
from cobra_ai.module.ontobj import *
from cobra_ai.module.utils import *
from cobra_ai.model.onto_vae import *
```

```python
# load ontobj
ontobj = Ontobj()
ontobj.load('/workspace/cobra_ai/data/GO/GO.ontobj')
```

```python
# load anndata (Kang PBMC data - log normalized)
adata = sc.read_h5ad('/workspace/cobra_ai/data/train_pbmc.h5ad')
```

```python
# create model
model = OntoVAE()
model.construct(
    ontobj=ontobj,
    adata=adata,
    root_layer_latent=True,
    linear_decoder=True
)
```

```python
# train the model
model.fit('/workspace/cobra_ai/test',   
                     lr=1e-4,                                 
                     kl_coeff=1e-4,                           
                     batch_size=128,                          
                     epochs=5)      
```

```python
# load the best model
model = OntoVAE.load('/workspace/cobra_ai/test')
```

```python
# get latent space embedding (stored in adata.obsm)
adata = model.to_latent(adata)
```

```python
# compute pathway activities (stored in adata.obsm)
adata = model.get_pathway_activities(adata)
```
